```python

#1. **Abstraction**: The process of hiding complex implementation details and showing only the essential features of an object. It focuses on 'what' an object does rather than 'how' it does it. This can be achieved using abstract classes and interfaces.
#2. **Encapsulation**: The bundling of data (attributes) and methods (functions) that operate on the data into a single unit or class. It restricts direct access to some of an object's components, which can prevent accidental modification of data. In Python, this is typically handled by convention (e.g., using a single underscore for 'protected' and double underscore for 'private' attributes).
#3. **Inheritance**: A mechanism that allows a new class (subclass/child class) to inherit properties (attributes and methods) from an existing class (superclass/parent class). This promotes code reusability and establishes an 'is-a' relationship between classes. Types include Single, Multilevel, Multiple, and Hybrid inheritance.
#4. **Polymorphism**: The ability of an object to take on many forms. In object-oriented programming, it allows methods to do different things depending on the object it is acting upon. This can be achieved through method overriding (same method name, different implementation in child class) and method overloading (not directly supported in Python in the traditional sense, but can be achieved using default arguments or `*args`/`**kwargs`).
```

### Example for Abstraction

```python
from abc import ABC, abstractmethod

# Abstract Base Class (ABC)
class Vehicle(ABC):
    def __init__(self, make, model):
        self.make = make
        self.model = model

    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def drive(self):
        pass

    def get_info(self):
        return f"Make: {self.make}, Model: {self.model}"

# Concrete class implementing Vehicle
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started quietly."

    def drive(self):
        return "Driving the car on the road."

# Another concrete class implementing Vehicle
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine roaring to life!"

    def drive(self):
        return "Riding the motorcycle on the highway."

# Creating objects
my_car = Car("Toyota", "Camry")
my_bike = Motorcycle("Harley-Davidson", "Iron 883")

print(f"Car Info: {my_car.get_info()}")
print(my_car.start_engine())
print(my_car.drive())

print(f"\nMotorcycle Info: {my_bike.get_info()}")
print(my_bike.start_engine())
print(my_bike.drive())
```

### Example for Encapsulation

```python
class BankAccount:
    def __init__(self, account_holder, initial_balance=0):
        self.account_holder = account_holder  # Public attribute
        self.__balance = initial_balance      # Private attribute by convention

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.__balance}")
        elif amount > self.__balance:
            print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        return self.__balance

# Create an account
account = BankAccount("Alice", 1000)

print(f"Account Holder: {account.account_holder}")
print(f"Initial Balance: ${account.get_balance()}")

account.deposit(500)
account.withdraw(200)
account.withdraw(1500) # Insufficient funds
account.deposit(-100) # Invalid deposit

# Trying to access private attribute directly (Python allows this, but it's discouraged)
# print(account.__balance) # This would raise an AttributeError because of name mangling
print(f"Current Balance (after operations): ${account.get_balance()}")
```

### Example for Inheritance

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        print("Some generic animal sound")

    def eat(self):
        print("Animal is eating.")

# Dog inherits from Animal (Single Inheritance)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "Dog") # Call parent class constructor
        self.breed = breed

    def make_sound(self): # Method Overriding
        print("Woof! Woof!")

    def fetch(self):
        print(f"{self.name} is fetching the ball.")

# GoldenRetriever inherits from Dog (Multilevel Inheritance)
class GoldenRetriever(Dog):
    def __init__(self, name, color):
        super().__init__(name, "Golden Retriever")
        self.color = color

    def make_sound(self):
        print("Happy bark!")

    def swim(self):
        print(f"{self.name} loves to swim.")

# Demonstrate
my_animal = Animal("Leo", "Lion")
my_dog = Dog("Buddy", "Labrador")
my_golden = GoldenRetriever("Goldie", "Golden")

print("Animal actions:")
my_animal.make_sound()
my_animal.eat()

print("\nDog actions:")
print(f"Name: {my_dog.name}, Species: {my_dog.species}, Breed: {my_dog.breed}")
my_dog.make_sound()
my_dog.eat()
my_dog.fetch()

print("\nGolden Retriever actions:")
print(f"Name: {my_golden.name}, Species: {my_golden.species}, Breed: {my_golden.breed}, Color: {my_golden.color}")
my_golden.make_sound()
my_golden.eat()
my_golden.fetch()
my_golden.swim()
```

### Example for Polymorphism

```python
class Bird:
    def fly(self):
        print("Birds can fly.")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly, they swim.")

class Airplane:
    def fly(self):
        print("Airplanes fly with engines.")

# Function demonstrating polymorphism
def make_it_fly(obj):
    obj.fly()

# Create objects
bird = Bird()
penguin = Penguin()
airplane = Airplane()

print("Demonstrating polymorphism:")
make_it_fly(bird)
make_it_fly(penguin)
make_it_fly(airplane)


# Another example: Method Overloading (achieved via default arguments or *args/**kwargs)
class Calculator:
    def add(self, a, b=None, c=None):
        if c is not None:
            return a + b + c
        elif b is not None:
            return a + b
        else:
            return a

calc = Calculator()
print(f"Addition (one arg): {calc.add(5)}")
print(f"Addition (two args): {calc.add(5, 10)}")
print(f"Addition (three args): {calc.add(5, 10, 15)}")
```

```python
# Error Handling in Python

#Error handling is a crucial part of writing robust software. In Python, errors and exceptional conditions are managed using `try-except` blocks. This mechanism allows you to gracefully handle unexpected situations in your code, preventing it from crashing.

#Key components of error handling:

#1.try block: This block contains the code that might raise an exception. If an error occurs within the `try` block, Python stops executing the rest of the code in that block and jumps to the `except` block.

#2.  **`except` block**: This block catches and handles specific exceptions that occur in the `try` block. You can specify the type of exception to catch (e.g., `ValueError`, `TypeError`, `ZeroDivisionError`), or use a general `except` to catch any exception. It's good practice to catch specific exceptions rather than broad ones.

#3.else block (optional): If provided, the code inside the `else` block is executed only if the `try` block completes successfully without raising any exceptions.

#4.finally block (optional): The code inside the `finally` block is always executed, regardless of whether an exception occurred in the `try` block or was handled by an `except` block. This is useful for cleanup operations, like closing files or releasing resources.

### Common Error Types:

#SyntaxError: Occurs when there's a grammatical mistake in the code (e.g., misspelled keyword, missing colon).
#NameError: Occurs when a variable or function is not defined.
#TypeError: Occurs when an operation is applied to an object of an inappropriate type (e.g., adding a string to an integer).
#ValueError: Occurs when a function receives an argument of the correct type but an inappropriate value (e.g., `int('hello')`).
#IndexError: Occurs when an index is out of range for a sequence (e.g., list, tuple).
#KeyError: Occurs when a dictionary key is not found.
#ZeroDivisionError: Occurs when division by zero is attempted.
```

### Example 1: Basic `try-except` for `ZeroDivisionError`

```python
def divide(a, b):
    try:
        result = a / b
        print(f"Result of division: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")

divide(10, 2)
divide(10, 0)
```

### Example 3: `try-except-else` block

```python
def process_file(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except IOError:
        print(f"Error: Could not read file '{filename}'.")
    else:
        print(f"File '{filename}' read successfully. Content length: {len(content)}")
        # print(content) # Uncomment to see content

# Create a dummy file for demonstration
with open('my_data.txt', 'w') as f:
    f.write('This is some test data.')

process_file('my_data.txt')
process_file('non_existent_file.txt')
```

### Example 4: `try-except-finally` block for cleanup

```python
def divide_and_cleanup(a, b):
    file = None
    try:
        file = open('log.txt', 'w')
        result = a / b
        file.write(f"Division result: {result}\n")
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: Division by zero occurred!")
        if file: file.write("Error: Division by zero\n")
    except TypeError:
        print("Error: Invalid types for division!")
        if file: file.write("Error: Invalid types\n")
    finally:
        if file:
            file.close()
            print("File 'log.txt' closed.")

# Test cases
divide_and_cleanup(10, 2)
print("\n")
divide_and_cleanup(10, 0)
print("\n")
divide_and_cleanup(10, 'a')
```

### Example 5: Handling multiple exceptions in a single `except` block

```python
def handle_errors(data, index):
    try:
        value = data[index]
        result = 10 / value
        print(f"Result: {result}")
    except (IndexError, ZeroDivisionError, TypeError) as e:
        print(f"An error occurred: {e} (Type: {type(e).__name__})")

my_list = [1, 2, 0, 4]

handle_errors(my_list, 1) # No error, result: 5.0
handle_errors(my_list, 4) # IndexError
handle_errors(my_list, 2) # ZeroDivisionError
handle_errors(my_list, 'a') # TypeError (if data is not string)
handle_errors("hello", 0) # No error, result: 10/h (TypeError at division)
```
