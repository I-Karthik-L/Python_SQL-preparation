```python
print("1. Creating Sets:")
# Creating an empty set
empty_set = set()
print(f"Empty set: {empty_set}")

# Creating a set with initial elements
my_set = {1, 2, 3, 4, 5}
print(f"Initial set: {my_set}")

# Creating a set from a list
list_data = [3, 4, 5, 6, 6, 7]
set_from_list = set(list_data)
print(f"Set from list: {set_from_list}") # Duplicates are automatically removed

print("\n2. Modifying Sets:")
# Adding elements
my_set.add(6)
my_set.add(1) # Adding an existing element has no effect
print(f"After adding 6: {my_set}")

# Adding multiple elements using update()
another_list = [7, 8]
my_set.update(another_list)
print(f"After updating with [7, 8]: {my_set}")

# Removing elements
my_set.remove(8) # Raises KeyError if element not found
print(f"After removing 8: {my_set}")

my_set.discard(10) # Does not raise an error if element not found
print(f"After discarding 10: {my_set}")

my_set.pop() # Removes and returns an arbitrary element
print(f"After popping an element: {my_set}")

my_set.clear() # Removes all elements
print(f"After clearing the set: {my_set}")

# Re-initialize my_set for further operations
my_set = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
set_c = {1, 2}

print("\n3. Set Operations:")
# Union: elements in either set
union_set = my_set.union(set_b)
# union_set = my_set | set_b # Alternative using operator
print(f"Union of {my_set} and {set_b}: {union_set}")

# Intersection: elements common to both sets
intersection_set = my_set.intersection(set_b)
# intersection_set = my_set & set_b # Alternative using operator
print(f"Intersection of {my_set} and {set_b}: {intersection_set}")

# Difference: elements in the first set but not in the second
difference_set = my_set.difference(set_b)
# difference_set = my_set - set_b # Alternative using operator
print(f"Difference ({my_set} - {set_b}): {difference_set}")

# Symmetric Difference: elements in either set, but not in both
symmetric_difference_set = my_set.symmetric_difference(set_b)
# symmetric_difference_set = my_set ^ set_b # Alternative using operator
print(f"Symmetric Difference ({my_set} ^ {set_b}): {symmetric_difference_set}")

print("\n4. Subset and Superset Checks:")
# issubset(): checks if all elements of one set are in another
is_subset = set_c.issubset(my_set)
# is_subset = set_c <= my_set # Alternative using operator
print(f"Is {set_c} a subset of {my_set}? {is_subset}")

# issuperset(): checks if one set contains all elements of another
is_superset = my_set.issuperset(set_c)
# is_superset = my_set >= set_c # Alternative using operator
print(f"Is {my_set} a superset of {set_c}? {is_superset}")

# Disjoint: checks if two sets have no elements in common
is_disjoint = my_set.isdisjoint({9, 10})
print(f"Are {my_set} and {{9, 10}} disjoint? {is_disjoint}")

is_disjoint_false = my_set.isdisjoint(set_b)
print(f"Are {my_set} and {set_b} disjoint? {is_disjoint_false}")

print("\n5. Other useful functions:")
# len(): number of elements in the set
print(f"Number of elements in {my_set}: {len(my_set)}")

# Iterating through a set
print("Elements in my_set:")
for item in my_set:
    print(item)

# Frozen set: immutable version of a set
frozen_set = frozenset([1, 2, 3])
print(f"Frozen set: {frozen_set}")
# frozen_set.add(4) # This would raise an AttributeError
```

```python
#WAP to find number of digits in a number:
n=int(input("Enter the number: "))
c=0
while n>0:
  n//=10
  c+=1
print("Number of digits: ",c)
```

```python
#WAP to find the sum of digits of a number:
n=int(input("Enter the number: "))
sum=0
while n>0:
  r=n%10
  sum+=r
  n//=10
print("Sum of digits: ",sum)
```

```python
#WAP to find the sum of square of digits of a number:
n=int(input("Enter the number: "))
sum=0
while n>0:
  r=n%10
  sum=r**2+sum
  n//=10
print("Sum of digits: ",sum)
```

```python
#WAP to check whether a given number is armstrong or not
n=int(input("Enter the number: "))
s=0
temp=n
l=len(str(n))
while n>0:
  r=n%10
  s+=r**l
  n//=10
if s==temp:
  print(f"{temp} is armstrong")
else:
  print(f"{temp} is not armstrong")
```

```python
print("Jumping Statements in Python:\n")

# 1. break statement
print("1. break statement:")
print("The 'break' statement terminates the loop entirely and transfers control to the statement immediately following the loop.")
for i in range(1, 10):
  if i == 5:
    print(f"  Breaking loop when i is {i}")
    break
  print(f"  Current number: {i}")
print("Loop finished (break example).\n")

# 2. continue statement
print("2. continue statement:")
print("The 'continue' statement skips the rest of the current iteration of the loop and proceeds to the next iteration.")
for i in range(1, 10):
  if i % 2 == 0: # Skip even numbers
    print(f"  Skipping even number: {i}")
    continue
  print(f"  Current odd number: {i}")
print("Loop finished (continue example).\n")

# 3. pass statement
print("3. pass statement:")
print("The 'pass' statement is a null operation; nothing happens when it executes. It is used as a placeholder when a statement is syntactically required but you do not want any command or code to execute.")

def my_function():
  pass # Placeholder for future code

print("  'pass' in a function definition:")
print("  def my_function():\n    pass")

print("  'pass' in a loop:")
for i in range(3):
  if i == 1:
    pass # Do nothing for i=1
    print("    'pass' executed for i = 1, but nothing happened.\n")
  else:
    print(f"    Processing number: {i}")
print("Loop finished (pass example).")
```

# Day 13 – Function Arguments and Searching

## Topics Covered
- Default Arguments
- Variable-Length Arguments (`*args`)
- Keyword Variable-Length Arguments (`**kwargs`)
- Linear Search

---





```python
def student(name,rollno=13):        #default arguments
  print("Name is: ",name)
  print("RollNo is: ",rollno)
student(rollno=12,name="Karthik")   #keyword arguments
```

```python
#*args,**kwargs
def one(*args):              #Normal arguments
  print(type(args))
  for i in args:
    print(i,end=" ")
one(1,2,3.14,True,"Name")
```

```python
def two(**kwargs):           #keyword arguments
  print(type(kwargs))
  for i,j in kwargs.items():
    print(i,j)
two(a=1,b=2)
```

```python
def onetwo(*args,**kwargs):
  print(type(kwargs))
  for i,j in kwargs.items():
    print(i,j)
  print(args)
onetwo(1,2,3,4,5,a=12,b=34)
```

```python
#SEARCHING
#1:Linear Searching

l=list(map(int,input("Enter the elements: ").split()))
k=int(input("Enter the number to be found: "))
for i in l:
  if i==k:
    print(f"Item found {i} at {l.index(i)} index")
    break
else:
  print("Item not found")
```

```python
# Binary Search = Divide and Conquer method
# Note: Binary search strictly requires a sorted list to work.

l1 = list(map(int, input("Enter the elements: ").split()))

# Sort the list to prepare it for binary search
l1.sort()
print(f"Sorted list being searched: {l1}")

k = int(input("Enter the value to be found: "))

u = len(l1) - 1
l = 0

while l <= u:
    mid = (l + u) // 2

    if l1[mid] < k:
        l = mid + 1
    elif l1[mid] > k:
        u = mid - 1
    else:
        # Simply return the 'mid' index which takes O(1) time
        print(f"Value found {l1[mid]} at index {mid}")
        break
else:
    print("Value not found")
```
