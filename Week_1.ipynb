## Type Conversion vs Type Casting

| Feature           | Type Conversion (Implicit)                                  | Type Casting (Explicit)                               |
| :---------------- | :---------------------------------------------------------- | :---------------------------------------------------- |
| **Definition**    | Automatic conversion of data from one type to another.      | User-defined conversion of data from one type to another. |
| **Initiation**    | Done by the compiler/interpreter automatically.             | Done by the programmer using specific syntax/functions. |
| **Safety**        | Generally safe, as it usually promotes to a 'wider' type.    | Can lead to data loss if converting to a 'narrower' type. |
| **Example (Python)**| Adding an integer to a float: `5 + 3.14` results in `8.14` (int converted to float). | Using `int()`, `str()`, `float()`, etc.: `int('10')` results in `10`. |
| **Use Case**      | When operations involve mixed data types (e.g., arithmetic). | When specific data types are required for functions or logic. |


## Global variables vs Local variables

| Feature           | Global Variables                                          | Local Variables                                             |
| :---------------- | :-------------------------------------------------------- | :---------------------------------------------------------- |
| **Definition**    | Declared outside any function.                            | Declared inside a function.                                 |
| **Scope**         | Accessible throughout the entire program.                 | Accessible only within the function where they are defined. |
| **Lifetime**      | Created when the program starts, destroyed when it ends.  | Created when the function is called, destroyed when it exits. |
| **Memory**        | Stored in a fixed location (data segment).                | Stored on the function call stack.                          |
| **Access**        | Can be read and modified from anywhere (with `global` keyword for modification inside functions). | Can only be accessed within their defining function.         |
| **Visibility**    | Visible to all functions.                                 | Not visible to other functions.                             |


## Python Data Types

Python has several built-in data types to categorize different kinds of values. Understanding these types is crucial as they determine what operations can be performed on a value.

| Data Type   | Description                                          | Example                             |
| :---------- | :--------------------------------------------------- | :---------------------------------- |
| `int`       | Integers (whole numbers, positive or negative)       | `10`, `-5`, `0`                     |
| `float`     | Floating-point numbers (numbers with decimal points) | `3.14`, `-0.5`, `2.0`               |
| `str`       | Strings (sequences of characters)                    | `'hello'`, `"Python"`, `'123'`    |
| `bool`      | Booleans (True or False)                             | `True`, `False`                     |
| `list`      | Ordered, mutable sequence of items                   | `[1, 2, 'a']`, `['apple', 'banana']`|
| `tuple`     | Ordered, immutable sequence of items                 | `(1, 2, 'a')`, `('red', 'green')`   |
| `dict`      | Unordered collection of key-value pairs              | `{'name': 'Alice', 'age': 30}`      |
| `set`       | Unordered collection of unique items                 | `{1, 2, 3}`, `{'a', 'b', 'c'}`      |


## Python Operators

Operators are special symbols or keywords that perform operations on one or more operands (values or variables).

| Operator Type        | Description                                                  | Examples                                         |
| :------------------- | :----------------------------------------------------------- | :----------------------------------------------- |
| **Arithmetic**       | Perform mathematical operations                              | `+`, `-`, `*`, `/`, `%`, `**`, `//`              |
| **Comparison**       | Compare two values and return a boolean result               | `==`, `!=`, `>`, `<`, `>=`, `<=`                |
| **Assignment**       | Assign values to variables                                   | `=`, `+=`, `-=`, `*=`, `/=`, `%=`, `**=`, `//=` |
| **Logical**          | Combine conditional statements                               | `and`, `or`, `not`                               |
| **Identity**         | Compare the memory locations of two objects                  | `is`, `is not`                                   |
| **Membership**       | Test if a sequence contains a specified value                | `in`, `not in`                                   |
| **Bitwise**          | Perform operations on individual bits of integers            | `&`,`^`, `~`, `<<`, `>>` ,`!`              |

```python
# --- Arithmetic Operators ---
print('--- Arithmetic Operators ---')
x = 15
y = 4

print(f'x + y = {x + y}')  # Addition
print(f'x - y = {x - y}')  # Subtraction
print(f'x * y = {x * y}')  # Multiplication
print(f'x / y = {x / y}')  # Division (returns float)
print(f'x % y = {x % y}')  # Modulus (remainder)
print(f'x ** y = {x ** y}') # Exponentiation
print(f'x // y = {x // y}') # Floor division (returns int)
```

```python
# --- Comparison Operators ---
print('\n--- Comparison Operators ---')
a = 10
b = 12

print(f'a == b is {a == b}') # Equal to
print(f'a != b is {a != b}') # Not equal to
print(f'a > b is {a > b}')   # Greater than
print(f'a < b is {a < b}')   # Less than
print(f'a >= b is {a >= b}') # Greater than or equal to
print(f'a <= b is {a <= b}') # Less than or equal to
```

```python
# --- Assignment Operators ---
print('\n--- Assignment Operators ---')
c = 5
print(f'Initial c = {c}')
c += 3 # c = c + 3
print(f'c += 3 results in c = {c}')
c -= 2 # c = c - 2
print(f'c -= 2 results in c = {c}')
c *= 4 # c = c * 4
print(f'c *= 4 results in c = {c}')
c /= 2 # c = c / 2
print(f'c /= 2 results in c = {c}')
```

```python
# --- Logical Operators ---
print('\n--- Logical Operators ---')
p = True
q = False

print(f'p and q is {p and q}') # Logical AND
print(f'p or q is {p or q}')   # Logical OR
print(f'not p is {not p}')     # Logical NOT
```

```python
# --- Identity Operators ---
print('\n--- Identity Operators ---')
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(f'list1 is list2 is {list1 is list2}')     # False (different memory locations)
print(f'list1 == list2 is {list1 == list2}')     # True (same values)
print(f'list1 is list3 is {list1 is list3}')     # True (same memory location)
print(f'list1 is not list2 is {list1 is not list2}') # True
```

```python
# --- Membership Operators ---
print('\n--- Membership Operators ---')
my_string = "hello world"
my_list = [10, 20, 30, 40]

print(f"'h' in my_string is {'h' in my_string}")     # True
print(f"'z' in my_string is {'z' in my_string}")     # False
print(f"20 in my_list is {20 in my_list}")           # True
print(f"50 not in my_list is {50 not in my_list}")   # True
```

```python
# --- Bitwise Operators ---
print('\n--- Bitwise Operators ---')
num1 = 10  # Binary: 1010
num2 = 4   # Binary: 0100

print(f'num1 & num2 = {num1 & num2}') # Bitwise AND
print(f'num1 | num2 = {num1 | num2}') # Bitwise OR
print(f'num1 ^ num2 = {num1 ^ num2}') # Bitwise XOR
print(f'~num1 = {~num1}')             # Bitwise NOT (one's complement)
print(f'num1 << 2 = {num1 << 2}')     # Left shift
print(f'num1 >> 2 = {num1 >> 2}')     # Right shift
```

```python
# Strings are sequences of characters, enclosed in single or double quotes.
string = "  Hello Python!  "
print(f"Original string:{string}")

# Common String Methods:

# 1. len(): Get the length of the string
print(f"Length: {len(string)}")

# 2. .strip(): Remove leading/trailing whitespace
stripped_string = string.strip()
print(f"Stripped: '{stripped_string}'")

# 3. .upper() / .lower(): Change case
print(f"Uppercase: '{string.upper()}'")
print(f"Lowercase: '{string.lower()}'")

# 4. .replace(old, new): Replace substrings
replaced_string = stripped_string.replace('Python', 'World')
print(f"Replaced: '{replaced_string}'")

# 5. .split(separator): Split string into a list
csv_data = "apple,banana,cherry"
print(f"Split by comma: {csv_data.split(',')}")
```

```python
# String Indexing: Accessing individual characters
# Strings are ordered. Each character has a position (index), starting from 0.
my_string = "Python"
print(f"Original string: '{my_string}'\n")

# Positive indexing: From the start (0, 1, 2...)
print(f"First character (index 0): {my_string[0]} ") # P
# Negative indexing: From the end (-1, -2...)
print(f"Last character (index -1): {my_string[-1]} \n") # n

# Slicing extracts a part of the string: string[start:end] (end is exclusive)
my_sentence = "Hello World"
print(f"Original sentence: '{my_sentence}'\n")

# Get 'Hello'
print(f"Slice from index 0 to 5: {my_sentence[0:5]} ")
# Get 'World'
print(f"Slice from index 6 to end: {my_sentence[6:]} ")
# Reverse the string
print(f"Reversed: {my_sentence[::-1]} ")
```

# Day 07 – Tuples and Dictionaries

### Topics Covered

- Tuple Operations

```python
print("### Understanding Tuples ###\n")

# What is a Tuple?
# A tuple is an ordered, immutable collection of items in Python.
# 'Immutable' means that once a tuple is created, you cannot change its elements.
# Tuples are defined by enclosing the elements in parentheses `()`.

# 1. Creating Tuples
print("1. Creating Tuples:")
my_tuple = (1, 2, 3, 'a', 'b')
print(f"  - Basic tuple: {my_tuple}")
single_element_tuple = (1,) # Note the comma is essential for a single-element tuple
print(f"  - Single element tuple: {single_element_tuple}")
another_tuple = 4, 5, 6 # Parentheses are optional if not ambiguous
print(f"  - Tuple without parentheses: {another_tuple}")
empty_tuple = ()
print(f"  - Empty tuple: {empty_tuple}")
print("\n")

# 2. Accessing Elements (Indexing and Slicing)
print("2. Accessing Elements (Indexing and Slicing):")
# Tuples are indexed like lists, starting from 0.
print(f"  - First element of {my_tuple}: {my_tuple[0]}")
print(f"  - Last element of {my_tuple}: {my_tuple[-1]}")

# Slicing works the same way as with strings and lists.
print(f"  - Slice from index 1 to 3 (exclusive) of {my_tuple}: {my_tuple[1:4]}")
print(f"  - Slice from beginning to index 2 (exclusive) of {my_tuple}: {my_tuple[:2]}")
print(f"  - Slice from index 3 to end of {my_tuple}: {my_tuple[3:]}")
print("\n")

# 3. Immutability of Tuples
print("3. Immutability of Tuples:")
# You cannot change individual elements of a tuple after creation.
# The following line would raise a TypeError:
# my_tuple[0] = 10
# print(f"  - Attempting to change an element: (will cause error if uncommented)")

# However, if a tuple contains mutable objects (like lists), those mutable objects can be changed.
mixed_tuple = (1, [2, 3], 'hello')
print(f"  - Original mixed tuple: {mixed_tuple}")
mixed_tuple[1].append(4) # Modifying the list inside the tuple
print(f"  - Mixed tuple after modifying inner list: {mixed_tuple}")
print("\n")

# 4. Common Functions and Operations on Tuples
print("4. Common Functions and Operations on Tuples:")
tuple_a = (1, 2, 3, 2, 4, 5)
tuple_b = (6, 7)

# len(): Returns the number of items in the tuple
print(f"  - Length of {tuple_a}: {len(tuple_a)}")

# count(): Returns the number of times a specified value occurs in a tuple
print(f"  - Count of '2' in {tuple_a}: {tuple_a.count(2)}")

# index(): Searches the tuple for a specified value and returns its position
print(f"  - Index of '3' in {tuple_a}: {tuple_a.index(3)}")
# print(tuple_a.index(9)) # This would raise a ValueError

# Concatenation (+): Joins two or more tuples
combined_tuple = tuple_a + tuple_b
print(f"  - Concatenation of {tuple_a} and {tuple_b}: {combined_tuple}")

# Repetition (*): Repeats the tuple elements
repeated_tuple = tuple_a * 2
print(f"  - Repetition of {tuple_a} (2 times): {repeated_tuple}")

# Membership (in, not in): Checks if an item exists in the tuple
print(f"  - Is '4' in {tuple_a}?: {'4' in tuple_a}")
print(f"  - Is '9' not in {tuple_a}?: {'9' not in tuple_a}")

# Iteration: Loop through tuple elements
print(f"  - Iterating through {tuple_a}:")
for item in tuple_a:
    print(f"    Item: {item}")

# min(), max(), sum(): For tuples containing numeric values
numeric_tuple = (10, 5, 20, 15)
print(f"  - Minimum value in {numeric_tuple}: {min(numeric_tuple)}")
print(f"  - Maximum value in {numeric_tuple}: {max(numeric_tuple)}")
print(f"  - Sum of values in {numeric_tuple}: {sum(numeric_tuple)}")

# Converting to other data types
list_from_tuple = list(tuple_a)
print(f"  - Converting {tuple_a} to a list: {list_from_tuple} (type: {type(list_from_tuple)})")
set_from_tuple = set(tuple_a)
print(f"  - Converting {tuple_a} to a set: {set_from_tuple} (type: {type(set_from_tuple)})")

# Converting from other data types
list_data = [10, 20, 30]
tuple_from_list = tuple(list_data)
print(f"  - Converting {list_data} (list) to a tuple: {tuple_from_list} (type: {type(tuple_from_list)})")
print("\n")
```

```python
print("### Understanding Dictionaries ###\n")

# What is a Dictionary?
# A dictionary is an unordered collection of data values, used to store data values like a map.
# Unlike other Data Types that hold only a single value as an element, a Dictionary holds 'key:value' pairs.
# Dictionaries are defined by enclosing the elements in curly braces `{}` with key-value pairs separated by colons `: `.
# Keys must be unique and immutable (e.g., strings, numbers, tuples). Values can be any data type.

# 1. Creating Dictionaries
print("1. Creating Dictionaries:")
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(f"  - Basic dictionary: {my_dict}")

another_dict = dict(brand='Ford', model='Mustang', year=1964)
print(f"  - Dictionary using dict() constructor: {another_dict}")

empty_dict = {}
print(f"  - Empty dictionary: {empty_dict}")

# Dictionary with mixed keys and values
mixed_keys_dict = {
    'name': 'Bob',
    1: 'one',
    (2, 3): 'tuple_key_value'
}
print(f"  - Dictionary with mixed keys/values: {mixed_keys_dict}")
print("\n")

# 2. Accessing Elements
print("2. Accessing Elements:")
# Elements are accessed by referring to their key name.
print(f"  - Access 'name' from {my_dict}: {my_dict['name']}")
print(f"  - Access 'year' from {another_dict}: {another_dict['year']}")

# Using .get() method for safer access (returns None if key not found, or a default value)
print(f"  - Access 'age' using .get() from {my_dict}: {my_dict.get('age')}")
print(f"  - Access non-existent key 'country' using .get() from {my_dict}: {my_dict.get('country')}")
print(f"  - Access non-existent key 'country' with default value from {my_dict}: {my_dict.get('country', 'Unknown')}")

# Accessing a non-existent key directly would raise a KeyError:
# print(my_dict['country']) # This would raise a KeyError if uncommented
print("\n")

# 3. Modifying Dictionaries (Adding, Updating, Deleting)
print("3. Modifying Dictionaries:")

# Adding a new key-value pair
my_dict['email'] = 'alice@example.com'
print(f"  - After adding 'email': {my_dict}")

# Updating an existing value
my_dict['age'] = 31
print(f"  - After updating 'age': {my_dict}")

# Deleting a key-value pair using del keyword
del my_dict['city']
print(f"  - After deleting 'city': {my_dict}")

# Using .pop() to remove a specific item and return its value
removed_name = my_dict.pop('name')
print(f"  - After popping 'name': {my_dict} (removed value: {removed_name})")

# Using .popitem() to remove the last inserted item (Python 3.7+)
# For older Python versions, it removes an arbitrary item.
popped_item = another_dict.popitem()
print(f"  - After popping an item from {another_dict}: {another_dict} (popped item: {popped_item})")

# Using .clear() to empty the dictionary
empty_dict.clear()
print(f"  - After clearing empty_dict: {empty_dict}")
print("\n")

# 4. Common Dictionary Methods and Operations
print("4. Common Dictionary Methods and Operations:")
student_grades = {'Math': 90, 'Science': 85, 'English': 92}

# len(): Returns the number of items (key-value pairs) in the dictionary
print(f"  - Length of {student_grades}: {len(student_grades)}")

# .keys(): Returns a view object that displays a list of all the keys
print(f"  - Keys: {student_grades.keys()}")

# .values(): Returns a view object that displays a list of all the values
print(f"  - Values: {student_grades.values()}")

# .items(): Returns a view object that displays a list of a dictionary's key-value tuple pairs
print(f"  - Items: {student_grades.items()}")

# .update(): Merges a dictionary with another dictionary or an iterable of key-value pairs
student_grades.update({'History': 88, 'Math': 95}) # Math gets updated
print(f"  - After updating: {student_grades}")

# Membership (in, not in): Checks if a key exists in the dictionary
print(f"  - Is 'Science' in {student_grades}?: {'Science' in student_grades}")
print(f"  - Is 'Art' not in {student_grades}?: {'Art' not in student_grades}")
print(f"  - (Note: 'in' checks keys, not values) Is 95 in {student_grades}?: {95 in student_grades}")
print("\n")

# 5. Iterating Through Dictionaries
print("5. Iterating Through Dictionaries:")
scores = {'John': 88, 'Jane': 92, 'Mike': 78}

print("  - Iterating through keys:")
for name in scores:
    print(f"    Name: {name}")

print("  - Iterating through keys (explicitly with .keys()):")
for name in scores.keys():
    print(f"    Name: {name}")

print("  - Iterating through values (with .values()):")
for score in scores.values():
    print(f"    Score: {score}")

print("  - Iterating through items (key-value pairs with .items()):")
for name, score in scores.items():
    print(f"    {name}'s score: {score}")
print("\n")

# 6. Dictionary Comprehension (similar to list comprehension)
print("6. Dictionary Comprehension:")
squared_numbers = {x: x**2 for x in range(1, 6)}
print(f"  - Squared numbers dictionary: {squared_numbers}")

# Example: Filtering a dictionary
high_scores = {k: v for k, v in scores.items() if v > 90}
print(f"  - High scores (above 90): {high_scores}")
print("\n")
```
