```python
#1: Lambda Function := Creates function in a single line
sq=lambda a:a*a   #Syntax (variable)= (keyword:lambda) (Argument):(what the arguments return)
print(sq(5))
```

```python
#WAP to create a lambda function to compute sum of two numbers
su=lambda a,b:a+b
print(su(8,9))
```

```python
#WAP to create a lambda function to find the largest of two numbers
l=lambda a,b:print(f"{a} is greater") if a>b else print(f"{b} is greater")
l(76,67)
```

```python
#WAP to create a lambda function to find whether a letter is vowel or not
s=lambda a:print(f"{a} is a vowel") if a.lower() in "aeiou" else print(f"{a} is not a vowel")
s("I")
```

```python
#2: Map Function

# The map() function applies a given function to each item of an iterable (like a list) and returns an iterator that yields the results. It's a core concept in functional programming for transforming data collections.

#Syntax:map(function, iterable)
```

```python
#WAP to input integers in a list using map and also find their square
l=list(map(int,input("Enter  the numbers: ").split()))
squared = list(map(lambda x: x**2,l))
print(squared)
```

```python
#WAP to find the sum respective numbers of two different list
l=list(map(int,input("Enter  the numbers of 1st list: ").split()))
m=list(map(int,input("Enter  the numbers of 2nd list: ").split()))
s=list(map(lambda a,b: a+b,l,m))
print(s)
```

---





```python
#3: Filter Function

# The filter() function constructs an iterator from elements of an iterable for which a function returns true. It's used to selectively choose elements from a collection based on a condition.

#Syntax: filter(function, iterable)

#Example:

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

```python
# WAP to return words that have more than 5 characters
l = input("Enter the words: ").split()
m=list(filter(lambda a:len(a)>5,l))
print(f"{m} are the words that have more than 5 characters")
```

```python
#WAP to create a list of odd numbers and even numbers from existing list
l=list(map(int,input("Enter  the numbers: ").split()))
odd=list(filter(lambda x:x%2!=0,l))
even=list(filter(lambda x:x%2==0,l))
print(f"{odd} are the odd numbers")
print(f"{even} are the even numbers")
```

```python
#4: List Comprehension

#List comprehension provides a concise way to create lists. It consists of brackets containing an expression followed by a `for` clause, then zero or more `for` or `if` clauses. The result is a new list resulting from evaluating the expression in the context of the `for` and `if` clauses which follow it.

#Syntax:[expression for item in iterable if condition]

numbers = [1, 2, 3, 4, 5]
squared_evens = [x**2 for x in numbers if x % 2 == 0]
print(squared_evens)
```

```python
#WAP to find squares of numbers from 1-10 in list format
l=[x**2 for x in range(1,11)]
print(l)
```

```python
l=list(map(int,input("Enter  the numbers: ").split()))
m=["even" if i%2==0 else "odd" for i in l]
print(m)
```

```python
## File Handling in Python

# File handling is an essential part of any application. It allows you to interact with files on your system, such as reading data from a file or writing data to a file.

# ### 1. Opening Files
# The `open()` function is used to open a file. It takes two primary arguments: the file path and the mode.
# Common modes:
# - `'r'` (read): Default mode. Opens a file for reading. Error if the file does not exist.
# - `'w'` (write): Opens a file for writing. Creates the file if it does not exist, or truncates (empties) the file if it exists.
# - `'a'` (append): Opens a file for appending. Creates the file if it does not exist. If the file exists, data is added to the end of the file.
# - `'x'` (create): Creates the specified file. Returns an error if the file exists.
# - `'t'` (text): Default mode. Opens in text mode.
# - `'b'` (binary): Opens in binary mode.

# Example: Open a file in write mode
# file = open('example.txt', 'w')
# file.close() # Always remember to close the file!

# ### 2. Writing to Files
# Use the `write()` method to write content to a file.

# Example: Writing to a file
# with open('my_file.txt', 'w') as f:
#     f.write('Hello, Python file handling!\n')
#     f.write('This is the second line.')

# ### 3. Reading from Files
# Use methods like `read()`, `readline()`, or `readlines()` to read content from a file.

# Example: Reading the entire file
# with open('my_file.txt', 'r') as f:
#     content = f.read()
#     print('Content of my_file.txt:')
#     print(content)

# Example: Reading line by line
# with open('my_file.txt', 'r') as f:
#     print('\nReading line by line:')
#     for line in f:
#         print(line.strip()) # .strip() removes leading/trailing whitespace, including newline characters

# ### 4. The `with` statement (Recommended)
# The `with` statement ensures that the file is properly closed after its block finishes, even if errors occur. This is the preferred way to handle files.

# Example of `with` statement for both writing and reading:
# Writing
with open('sample.txt', 'w') as f:
    f.write('First line.\n')
    f.write('Second line.')

# Reading
with open('sample.txt', 'r') as f:
    data = f.read()
    print('Content of sample.txt:')
    print(data)

# Appending
with open('sample.txt', 'a') as f:
    f.write('\nThird line (appended).')

# Reading after appending
with open('sample.txt', 'r') as f:
    updated_data = f.read()
    print('\nContent after appending:')
    print(updated_data)
```
